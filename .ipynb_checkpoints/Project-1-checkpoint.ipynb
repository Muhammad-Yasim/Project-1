```python
#for numbers in data['users'][0]['friends']:
def friends(friends_list):
    names=[]
    for f in friends_list:
        for user in data['users']:
            if f==user['id']:
                names.append(user['name'])
    if len(names)==0:
        return "No friends"
    elif len(names)==1:
        return names[0]
    elif len(names)==2:
        return " and ".join(names)
    else:
        return ",".join(names[:-1]) + " and " + names[-1]
    #return ",".join(names)
    #return ",".join(map(str,friends_list))         #Because cant directly use function in f-string---this is sued without creating list
```

```python
def liked_pages(like_pages_list):
    pages=[]
    for f in like_pages_list:
        for page in data['pages']:
            if f==page['id']:
                pages.append(page['name'])
    if len(pages)==0:
        return "No pages liked"
    elif len(pages)==1:
        return pages[0]
    elif len(pages)==2:
        return " and ".join(pages)
    else:
        return ",".join(pages[:-1]) + " and " + pages[-1]
```

```python

with open("data.json","r") as file_reader : 
    data = json.load(file_reader)
    
print(data)
print(type(data))
```

```python
print("-USERS INFORMATION-")
for user in data['users']:
    print(f"{user['id']}. {user['name']} is friends with {friends(user['friends'])} . Liked pages are { liked_pages(user['liked_pages']) }  ")
    
```
