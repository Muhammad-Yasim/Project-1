```python
def liked_pages(like_pages_list):
    pages=[]
    for f in like_pages_list:
        for page in data['pages']:
            if f==page['id']:
                pages.append(page['name'])
    if len(pages)==0:
        return "No pages liked"
    elif len(pages)==1:
        return pages[0]
    elif len(pages)==2:
        return " and ".join(pages)
    else:
        return " , ".join(pages[:-1]) + " and " + pages[-1]
```

```python
def pages_you_may_like(user_id,data):
    
    user_pages={}
    
    for user in data['users']:
        user_pages[user['id']]=set(user['liked_pages'])    #Dcitionary

    user_liked_pages=user_pages[user_id]      #Set
    
    page_suggestions={}
    
    for other_user,pages in user_pages.items():      #using 2 arguments bcz we have dictionary i.e KEY:VALUE or USER-ID:PAGES LIKED   in   user_pages.items()
        if other_user!=user_id:
            shared_pages=user_liked_pages.intersection(pages)
            for page in pages:
                if page not in user_liked_pages:
                    page_suggestions[page]=page_suggestions.get(page,1) + len(shared_pages)    #.get is used to enter new value if not available with default key value.If already there,then ignores the default value and rewrite ie
                    
    #Sorting
    sorted_pages=sorted( page_suggestions.items() , key=lambda x:x[1] , reverse=True )       #It will return a list of tuples
    
    return [i for i,j in sorted_pages]
    
```

```python
data=json.load(open("data3.json"))
```

```python
for user in data['users']:
    page_recommendations=pages_you_may_like(user['id'],data)
    print(f"{user['name']} may Like : {liked_pages(page_recommendations)}\n")
```
