```python
import json
```

```python

with open("data.json","r") as file_reader : 
    data = json.load(file_reader)
    
print(data)
print(type(data))
```

```python
print("-USERS INFORMATION-")
for user in data['users']:
    print(f"{user['id']}. {user['name']} is friends with {user['friends']} , liked pages are {user['liked_pages']}  ")
    
```

```python
print("-PAGES INFORMATION-")
for page in data['pages']:
    print(f" Page # {page['id']} is about  -> {page['name']}")
```
