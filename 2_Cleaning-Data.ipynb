```python
import json
```

```python
#Function for cleaning data
def cleaned_data(data):
    #Removing users wiith no name
    data['users']=[user for user in data['users'] if user['name'].split() ]
    
    #Removing duplicate values
    for user in data['users']:
        user['friends']=list(set(user['friends']))    #Converting to set to remove duplicates,then again into list)
        
    #Removing inactive users
    data['users']=[user for user in data['users'] if user['friends'] or user['liked_pages']]

    #Removing Duplicate Pages
    unique_pages={}    #Dictionary
    for page in data['pages']:
        unique_pages[page['id']]=page    #key : value   --- we know that there can only be 1 key at a tiome,no duplicate keys,so duplicate one will be rewritten
    data['pages']=list(unique_pages.values())
    
    return data
```

```python
#Friends list
def friends(friends_list):
    names=[]
    for f in friends_list:
        for user in data['users']:
            if f==user['id']:
                names.append(user['name'])
    if len(names)==0:
        return "No friends"
    elif len(names)==1:
        return names[0]
    elif len(names)==2:
        return " and ".join(names)
    else:
        return ",".join(names[:-1]) + " and " + names[-1]
    

#Liked-Pages list
def liked_pages(like_pages_list):
    pages=[]
    for f in like_pages_list:
        for page in data['pages']:
            if f==page['id']:
                pages.append(page['name'])
    if len(pages)==0:
        return "No pages liked"
    elif len(pages)==1:
        return pages[0]
    elif len(pages)==2:
        return " and ".join(pages)
    else:
        return ",".join(pages[:-1]) + " and " + pages[-1]
```

```python
data=json.load( open("data2.json") )   #Loading data from file
```

```python
print("---PRINTING BEFORE CLEANING---".center(60))
print("-USERS INFORMATION-")
for user in data['users']:
    print(f" {user['id']}. {user['name']} is friends with {friends(user['friends'])} . Liked pages are { liked_pages(user['liked_pages']) }  ")
    
print("\n-PAGES INFORMATION-")
for page in data['pages']:
    print(f" Page # {page['id']} is about  -> {page['name']}")
```

```python
data=cleaned_data(data)    #Calling function
json.dump(data,open("data2(cleaned).json","w"),indent=4)    #Writing data into file
print("Data Cleaned Written successfully")  
```
